```python
import numpy as np
import matplotlib.pyplot as plt

# Your dataset
sensor = np.array([616.6, 494.3, 387])
moisture = np.array([10.0, 7.7, 3.5])

# Create best-fit line (linear regression)
a, b = np.polyfit(sensor, moisture, 1)

# Generate line values
x_line = np.linspace(380, 620, 100)
y_line = a * x_line + b

# Plot
plt.figure(figsize=(8, 5))
plt.scatter(sensor, moisture, label="Data points")
plt.plot(x_line, y_line, label=f"Fit: y = {a:.4f}x + {b:.4f}")

# Labels and title
plt.xlabel("Sensor Value")
plt.ylabel("Soil Moisture (%)")
plt.title("Soil Moisture Calibration")
plt.legend()

plt.grid()
plt.show()

# Print equation
print(f"Calibration Equation: y = {a:.4f}x + {b:.4f}")
```
